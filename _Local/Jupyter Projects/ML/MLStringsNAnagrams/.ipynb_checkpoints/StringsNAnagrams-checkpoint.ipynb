```python
def getNSubstrings(s, n):
    nSubstrings = set(())
    for i in range(len(s) - n + 1):
        nSubstrings.add(s[i:i+n])
    return nSubstrings

def getNAnagrams(s1, s2, n):
    nSubstrings1 = getNSubstrings(s1, n)
    nSubstrings2 = getNSubstrings(s2, n)
    nAnagrams = [nSubstring for nSubstring in nSubstrings1 if nSubstring in nSubstrings2]
    return nAnagrams
```

```python
s1 = input("Read first string: ")
s2 = input("Read second string: ")
n = int(input("Read n: "))

nAnagrams = getNAnagrams(s1, s2, n)
print("Number of " + str(n) + "-Anagrams of " + s1 + " and " + s2 + ": " + str(len(nAnagrams)))
print(nAnagrams)
```
