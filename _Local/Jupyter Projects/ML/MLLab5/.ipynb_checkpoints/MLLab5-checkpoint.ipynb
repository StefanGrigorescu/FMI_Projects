```python
from sklearn import svm
# modelul bag of words transforma cuvintele din stringuri in date numerice

# pt un document, obtinem un dictionar de cuvinte hash:cuvant

# in ex nostru avem mai multe documente. Extragem multimea de cuvinte {w1, w2, w3}. Atribuim id-uri numerice
# pt fiecare document, facem un vocabular (o lista de frecventa cu cate aparitii au cuvintele in document)

class BOW:
    def __init__(self):
        self.vocabulary = {}
        self.words = []
        self.vocabularyLen = 0
        
    def buildVocabulary(self, data):
        for document in data:
            for word in document:
                if not word in self.vocabulary.keys():
                    self.vocabulary[word] = len(self.vocabulary)
                    self.words.append(word)
        self.vocabularyLen = len(self.vocabulary)

    def getFeatures(self, data):
        features = np.zeros(len(data), self.vocabularyLen)
        for docId, document in enumerate(data):
            for word in document:
                if word in self.vocabulary:
                    features[docId, self.vocabulary[word]] += 1
        return features
    
```
