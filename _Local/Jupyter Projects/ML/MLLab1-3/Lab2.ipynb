```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
train_images = np.loadtxt('data/train_images.txt')
train_labels = np.loadtxt('data/train_labels.txt', 'int')

image1 = train_images[0, :]
image1 = np.reshape(image1, (28, 28))
plt.imshow(image1.astype(np.uint8), cmap ='gray')
plt.show()

test_images = np.loadtxt('data/test_images.txt')
test_labels = np.loadtxt('data/test_labels.txt', 'int')

```

```python
bins = np.linspace(start=0, stop=255, num=4)  # returneaza intervalele
print(bins)
x_to_bins = np.digitize(train_images, bins)   # returneaza pentru fiecare element intervalul
                                              # corespunzator
                                              # Atentie! In cazul nostru indexarea elementelor va
                                              # incepe de la 1, intrucat nu avem valori < 0
print(x_to_bins)
```

```python
#exercitiul 2
num_bins = 5
bins = np.linspace(0, 255, num_bins)
def values_to_bins(matrice, bins):
    x = np.digitize(matrice, bins)
    x = x-1
    return x

x_train = values_to_bins(train_images, bins)
x_test = values_to_bins(test_images, bins)
print(x_train)
print(x_test)
```

```python
#exercitiul3
from sklearn.naive_bayes import MultinomialNB
naive_bayes_model = MultinomialNB()
naive_bayes_model.fit(x_train, train_labels)
naive_bayes_model.score(x_test, test_labels)
```
