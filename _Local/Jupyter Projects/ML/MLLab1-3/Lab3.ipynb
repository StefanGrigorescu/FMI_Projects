```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#ex 1 + 2 + 3
class KnnClassifier:
    def __init__(self, train_images, train_labels):
        self.train_images = train_images
        self.train_labels = train_labels
    def classify_image(self, test_image, num_neighbors = 3, metric ='l2'):
        if metric == 'l2':
            distances = np.sqrt(np.sum((self.train_images-test_image)**2, axis = 1))
        else:
            distances = np.sum(abs(self.train_images - test_image))
            
        sort_indexes = np.argsort(distances)
    
        images_neighbors = self.train_labels[sort_indexes]
        images_neighbors = images_neighbors[:num_neighbors]
    
        labels_count = np.bincount(images_neighbors)
        
        return np.argmax(labels_count)
    def ex3(self, test_images, num_neighbours = 3, metric = 'l2'):
        num_test_images = test_images.shape[0]
        predicted_labels = np.zeros(num_test_images)
        
        for i, test_image in enumerate(test_images):
            predicted_labels[i] = self.classify_image(test_image, num_neighbours, 'l2')
        
        return predicted_labels
    
    def accuracy(self, y_pred, y_true):
        return (y_pred == y_true).mean()
            
                    
        
```

```python
train_images = np.loadtxt('data/train_images.txt')
train_labels = np.loadtxt('data/train_labels.txt', 'int')
test_images = np.loadtxt('data/test_images.txt')
test_labels = np.loadtxt('data/test_labels.txt', 'int')

# train_images.shape = (num_samples, num_features),
# test_image.shape = (1, num_features) 

classifier = KnnClassifier(train_images, train_labels)


#for imagine in test_images:
    #print(classifier.classify_image(imagine, 3, 'l2'))
    
#print(classifier.accuracy(classifier.ex3(test_images, 3, 'l2'), test_labels))

#ex4

acurateti = []
nrVecini = [1,3,5,7,9]
for nr in nrVecini:
    acurateti += classifier.accuracy(classifier.ex3(test_images, nr, 'l2'), test_labels)
    np.array(acurateti)
    
#continua
```
