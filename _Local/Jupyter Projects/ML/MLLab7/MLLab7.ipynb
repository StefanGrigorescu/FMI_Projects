```python
# eroare = y - yCaciula = y - f(X*W+b)



# Obs: folosim derivata ca sa gasim puncte de minim global
# magnitudinea = gradientul =(not) derivata partiala in fct de o componenta = E(W, b) 
# Derivata patiala a loss in fct de ponderi 
# ObS: mergem in directia opusa semnului gradientului; gradientul 

# deci
# y = eticheta lui X
# yCaciula = predictia (dupa formula de mai sus)

# Alg Windrow-Hoff
# Loss  = (yCaciula - y)^2 / 2
# Derivata partiala a fct de Loss in functie de W   = d Loss / d W   (notez cu derivata)
# In cazul nostru,     d Loss = derivata(fct de doua randuri de mai sus) = 2 * (X*W+b-y) / 2 * X = (ycaciula - y) * X

# Actualizarea W:   W = W - niu (yCaciula - y) * X, unde niu este rata de invatare

 
```

```python
import numpy as np

# Alg WIndrow--Hoff
X = np.array([[0,0],[0,1], [1,0],[1,1]])
y = np.array([-1, 1, 1, 1])
W = np.zeros((2))
b = 0
numEpochs = 50
learningRate = 0.1
numSamples = X.shape[0]
# fct de activare este fct identitate in cazul nostru

```

```python
def compute_y(x, W, bias): 
    # dreapta de decizie
    # [x, y] * [W[0], W[1]] + b = 0
    return (-x * W[0] - bias) / (W[1] + 1e-10)


def plot_decision_boundary(X, y , W, b, current_x, current_y): 
    x1 = -0.5 
    y1 = compute_y(x1, W, b) 
    x2 = 0.5 
    y2 = compute_y(x2, W, b)
    # sterge continutul ferestrei
    plt.clf()
    # ploteaza multimea de antrenare
    color = 'r'
    if(current_y == -1):
        color = 'b'
    plt.ylim((-1, 2))    
    plt.xlim((-1, 2))  
    plt.plot(X[y == -1, 0], X[y == -1, 1], 'b+')
    plt.plot(X[y == 1, 0], X[y == 1, 1], 'r+')
    # ploteaza exemplul curent
    plt.plot(current_x[0], current_x[1], color+'s')
    # afisarea dreptei de decizie
    plt.plot([x1, x2] ,[y1, y2], 'black')
    plt.show(block=False)
    plt.pause(0.3)
```

```python
from sklearn.utils import shuffle

for e in range(numEpochs):
    X, y = shuffle(X, y) # 3.a  # epoca
    for i in range(numSamples):
        yHat = np.dot(X[i][:], W) + b   # predictie
        loss = ((yHat - y[i])**2)/2
        W = W - learningRate * (yHat - y[i] * X[i][:])
        b = b - learningRate * (yHat - y[i])
        plot_decision_boundary(X, y, W, b, X[i][:], y[i])

accuracy = (np.sign(np.dot(X, W) + b) == y).mean()
    
```
