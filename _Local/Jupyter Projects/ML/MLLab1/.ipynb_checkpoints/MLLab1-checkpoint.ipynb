```python
# Ex laborator:

# 1/ a)
images = np.zeros((9,400,600))  # 9 poze, cu dimensiunile 400x600
# puteam sa pun si (9, 400*600)  daca erau ravel (sau cum ii zicea, gen desfasurate), dar ele sunt ca matrice

number = 9
for i in range(number):
    image = np.load('images/car_{}.npy'.format(i))
    images[i] = image
```

```python
# b) 
print(np.sum(images, axis = (0,1,2)))   # mi-a dat raspunsul corect, 269782418.0, conform celorlaltora
```

```python
#c) 
print(np.sum(images, axis=(1,2)))
```

```python
#d) 

print(np.argmax(np.sum(images,axis=(1,2))))   # raspunsul corect este 1, mi-a dat ok
```

```python
# e)
meanImg = np.mean(images, axis = 0)
print(meanImg)

# deci ideea e ca tu practic tai axele pe care le dai in axis, ca aplici o functie de grup pe ele. 
# Asa poti tine minte cum funcitoneaza
```

```python
from skimage import io
io.imshow(meanImg.astype(np.uint8))  # int8 : val intre 0 si 255
# default erau floaturi
io.show()

# Obs: CV2 si openCV sunt librarii cu care pot transforma culorile in grey scale
```

```python
# f) 
devStandard = np.std(images)
print(devStandard)
```

```python
# g) 
meanImgMat = np.full((9, 400, 600), meanImg)

imagesModif = images - meanImgMat

devStdImgMat = np.full((9, 400, 600), devStandard)

imagesModif = imagesModif // devStdImgMat

for imgModif in imagesModif:
    io.imshow(imgModif.astype(np.uint8))  # uint8 : val intre 0 si 255
    io.show()
```
