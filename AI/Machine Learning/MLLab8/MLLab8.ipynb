```python
import numpy as np
from sklearn.linear_model import Perceptron # importul clasei

# threshold = 1e-5
X = np.loadtxt('./data/3d-points/x_train.txt')
y = np.loadtxt('./data/3d-points/y_train.txt', 'int')

#train perceptron

perceptronModel = Perceptron(eta0 = 0.00000001)
perceptronModel.fit(X, y)

print('accuracy: ' + str(perceptronModel.score(X, y)))
print('ponderi :' + str(perceptronModel.coef_))
print('bias ' + str(perceptronModel.intercept_))
```

```python
from mpl_toolkits import mplot3d
import numpy as np
import matplotlib.pyplot as plt
def plot3d_data_and_decision_function(X, y, W, b):
    ax = plt.axes(projection='3d')
    # create x,y
    xx, yy = np.meshgrid(range(10), range(10))
    # calculate corresponding z
    # [x, y, z] * [W[0], W[1], W[2]] + b = 0
    zz = (-W[0] * xx - W[1] * yy - b) / W[2]
    ax.plot_surface(xx, yy, zz, alpha=0.5)
    ax.scatter3D(X[y == -1, 0], X[y == -1, 1], X[y == -1, 2], 'b');
    ax.scatter3D(X[y == 1, 0], X[y == 1, 1], X[y == 1, 2], 'r');
    plt.show()


```

```python
# continuare lab precedent
def sigmoid(x):
    return 1.0/(1.0) + np.exp(-1.0 * x)
    
def tanhDerived(x):
    return 1.0 - np.tanh(x)**2

# yCaciula= sigmoid(tanh(X*W1+b1)*W2 + b2))
# yCaciula = 

def forward(X, W1, b1, W2, b2, numSamples):
    z1 = X.dot(W1) + b1
    # activationFunction 
    a1 = np.tanh(z1)
    z2 = a1.dot(W2) + b2
    a2 = sigmoid(z2)
    return z1, a1, z2, a2

def backward(z1, a1, z2, a2, W2, X, y, numSamples):
    dz2 = a2 - y # start calculating the slope of the loss function with respect to W2
    dW2 = (1/numSamples) * np.dot(a1.T, dz2)
    da1 = np.dot(dz2, W2.T)
    dz1 = np.multiply(da1, tanhDerived(z1))
    dW1 = (1/numSamples) * np.dot(X.T, dz1)
    db1 = (1/numSamples) * np.sum(dz1, axis = 0)
    return dW1, db1, dW2, db2

X = np.array([[0,0], [0, 1], [1, 0], [1,1]])    
y = np.expand_dims(np.array([0,1,0,1]), 1)
numHiddenLayers = 5 # asta e de fapt cati neuroni sunt pe singurul nostru strat ascuns, notatia nu e inspirata
numOutputNeurons = 1 # cati neuroni sunt pe output layer
W1 = np.random.normal(0, 1, (2, numHiddenLayers))
b1 = np.zeros(numHiddenLayers)
W2 = np.random.normal(0, 1, (numHiddenLayers, numOutputNeurons)) # 0 si 1 sunt media si deviatia standard (pt ca e distributia normala)
b2 = np.zeros(numOutputNeurons)
numSamples = X.shape[0]
numEpochs= 50
learningRate = 0.5
    
    
    
```
