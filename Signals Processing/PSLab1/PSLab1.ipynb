```python
import numpy as np
import matplotlib.pyplot as plt
import random

def getSamples(tList, fct):
    return [fct(t) for t in tList]

def getSamples1(tList, F, fct):
    return [fct(t, F) for t in tList]

def plot(timeAxis, fct):
    plt.stem(timeAxis, getSamples(timeAxis, fct))
    
def plot1(timeAxis, F, fct):
    plt.stem(timeAxis, getSamples1(timeAxis, F, fct))
```

```python
def x(t):
    return np.cos(520*np.pi*t + np.pi/3)

def y(t):
    return np.cos(280*np.pi*t - np.pi/3)

def z(t):
    return np.cos(120*np.pi*t + np.pi/3)

def semnalSin(t, F):
    return np.sin(2*F*np.pi*t + np.pi/3)

def semnalSawtooth(t, F):
    return np.mod(2*F*np.pi*t + np.pi/3, 2*np.pi)

def semnalSquare(t, F):
    return np.sign(semnalSin(t,F))


discreteTimeAxis = np.linspace(0, 0.03, 60)

# intitial stage:
# noSamplesPerSecond = 60 * 1/0.03 = 2000

# new stage:
# F = 200 Hz (= noSamplesPerSecond)
# noSamplesPer0.03Seconds = 200 * 0.03/1 = 6

discreteTimeAxis1 = np.linspace(0, 0.03, 6)
```

```python
# f

def getSignalF():
    signalF = np.eye(128, dtype=float)
    for i in range(len(signalF)):
        for j in range(len(signalF[i])):
            signalF[i][j] = float(signalF[i][j] * random.randint(0,9))
    return signalF

signalF = getSignalF()
plt.imshow(signalF)
```
